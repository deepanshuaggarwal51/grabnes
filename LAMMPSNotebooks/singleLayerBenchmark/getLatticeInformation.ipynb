```python
a1 = np.array([19.600000000  ,  12.730573436 ,  0.000000000])
a2 = np.array([-1.225000000 ,   23.339384632,  0.000000000])
a3 = np.array([0.0 , 0.0 , 40.40453015592109])

print("------------------ Initial lattice vectors ------------------")
print("a1 = ", a1)
print("a2 = ", a2)
print("a3 = ", a3)
print("norm(a1) = ", np.linalg.norm(a1))
print("norm(a2) = ", np.linalg.norm(a2))
print("\n")

zero = np.array([0, 0, 0])

#newline(zero, a1)
#newline(zero, a2)

plt.figure(figsize=(5,5))

plt.plot([zero[0], a1[0]], [zero[1], a1[1]], 'bo-')
plt.plot([zero[0], a2[0]], [zero[1], a2[1]], 'bo-')

plt.xlim(-10,140)
plt.ylim(-10,140)


e1 = np.array([1.0, 0.0, 0.0])

phi = -np.arccos(np.dot(e1, a1)/(np.linalg.norm(e1)*np.linalg.norm(a1)))

rad = 180/np.pi

# print(phi)

R = np.array([[np.cos(phi), -np.sin(phi), 0.0],
            [np.sin(phi), np.cos(phi), 0.0],
             [0.0, 0.0, 1.0]])

a1Rot = np.matmul(R,a1)
a2Rot = np.matmul(R,a2)
a3Rot = np.matmul(R,a3)

#print("check: ", np.linalg.norm(a1Rot))

print("------------------ Rotated cell lattice vectors - phi = " + str(phi*rad) + " ------------------")
print("-------------------- a1Rot must be aligned in the positive x direction ------------------------")
print("------------------ a2Rot must be in the xy plane with positive y component --------------------")
print("--------------------------- a3Rot must have a positive z component ----------------------------")

print("a1Rot: ", a1Rot)
print("a2Rot: ", a2Rot)
print("a3Rot: ", a3Rot)
print("norm(a1) = ", np.linalg.norm(a1Rot))
print("\n")

plt.plot([zero[0], a1Rot[0]], [zero[1], a1Rot[1]], 'ro--')
plt.plot([zero[0], a2Rot[0]], [zero[1], a2Rot[1]], 'ro--')

a = np.linalg.norm(np.matmul(R,a1))
b = np.linalg.norm(np.matmul(R,a2))
c = np.linalg.norm(np.matmul(R,a3))


#gamma = np.arccos(np.dot(a1, a2)/(np.linalg.norm(a1)*np.linalg.norm(a2)))
#print("gamma: ", gamma)
gamma = np.arccos(np.dot(a2Rot, a1Rot)/(np.linalg.norm(a2Rot)*np.linalg.norm(a1Rot)))
#print("gamma: ", gamma)
alpha = np.pi/2.0
beta = np.pi/2.0
#newline(zero, a)
#newline(zero, b)

print("------------------ Conventional triclinic cell notation ------------------")
print("a= ", a)
print("b= ", b)
print("c= ", c)
print("alpha: ", alpha*rad)
print("beta: ", beta*rad)
print("gamma: ", gamma*rad)
print("\n")



#print("abc:", a, b, c)

lx = a
xy = b*np.cos(gamma)
xz = c*np.cos(beta)
ly = np.sqrt(b**2 - xy**2)
yz = (b*c*np.cos(alpha) - xy*xz)/(ly)
lz = np.sqrt(c**2 - xz**2 - yz**2)
print("------------------ Parameters to be used by LAMMPS ------------------")
print("lx = ", lx)
print("ly = ", ly)
print("lz = ", lz)
print("xy = ", xy)
print("xz = ", xz)
print("yz = ", yz)
print("\n")

if (xy>lx/2):
    print("Choose other xy, it is too tilted")
    xy = xy-lx
    print("new xy = ", xy)
```
